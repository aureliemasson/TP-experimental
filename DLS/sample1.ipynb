```python
import pandas as pd
import matplotlib.pyplot as plt

# Charger le fichier Excel
file_path = "sample 1 all datas.xlsx"  # Mets le bon chemin ici
df = pd.read_excel(file_path, sheet_name="Feuil1")

# Extraction des colonnes pertinentes
size_col = df.columns[0]  # Première colonne de taille
intensity_col = df.columns[1]  # Deuxième colonne d'intensité

# Fonction pour tracer le graphe spécifique
def plot_size_intensity():
    plt.figure(figsize=(8, 6))
    mask = df[size_col] <= 400 # Filtrer pour arrêter à 400
    plt.plot(df[size_col].loc[mask], df[intensity_col].loc[mask], marker='o', linestyle=' ', label='Test 1')
    plt.xlabel("Taille (d.nm)")
    plt.ylabel("Intensité (%)")
    plt.title("Taille vs Intensité (Premières colonnes)")
    plt.legend()
    plt.grid(True)
    plt.show()

# Tracer le graphe spécifique
plot_size_intensity()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Charger le fichier Excel
df = pd.read_excel('sample 1 all datas vrai.xlsx')

# Sélectionner les colonnes pertinentes pour les échantillons
df_samples_1 = df[['Size (d.nm) - Sample 1 [Steady state] test1', 'Intensity (Percent) - Sample 1 [Steady state] test1']]
df_samples_2 = df[['Size (d.nm) - Sample 1 [Steady state] test2', 'Intensity (Percent) - Sample 1 [Steady state] test2']]
df_samples_3 = df[['Size (d.nm) - Sample 1 [Steady state] test3', 'Intensity (Percent) - Sample 1 [Steady state] test3']]

# Nettoyer les données en supprimant les lignes où il manque des valeurs
df_samples_1 = df_samples_1.dropna()
df_samples_2 = df_samples_2.dropna()
df_samples_3 = df_samples_3.dropna()

# Tracer les courbes
plt.figure(figsize=(10, 6))

# Courbes des trois échantillons
plt.plot(df_samples_1.iloc[:, 0], df_samples_1.iloc[:, 1], label="Sample 1 (Test 1)", color='b', linewidth=2)
plt.plot(df_samples_2.iloc[:, 0], df_samples_2.iloc[:, 1], label="Sample 1 (Test 2)", color='r', linewidth=2)
plt.plot(df_samples_3.iloc[:, 0], df_samples_3.iloc[:, 1], label="Sample 1 (Test 3)", color='g', linewidth=2)

# Définir les limites de l'axe des x pour qu'il s'arrête à 400 nm
plt.xlim(0, 400)

# Ajouter les titres, les labels et la légende
plt.xlabel("Size (d.nm)")
plt.ylabel("Intensity (%)")
plt.title("Intensity vs. Size for Sample 1 (Steady state)")
plt.legend()

# Afficher le graphique
plt.grid(True)
plt.show()

```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Charger le fichier Excel
df = pd.read_excel('sample 1 all datas vrai.xlsx')

# Sélectionner les colonnes pertinentes pour les échantillons
df_samples_1 = df[['Time (µs) - Sample 1 [Steady state] test1', 'Correlation Coefficient (g₂-1) - Sample 1 [Steady state] test1']]
df_samples_2 = df[['Time (µs) - Sample 1 [Steady state] test2', 'Correlation Coefficient (g₂-1) - Sample 1 [Steady state] test2']]
df_samples_3 = df[['Time (µs) - Sample 1 [Steady state] test3', 'Correlation Coefficient (g₂-1) - Sample 1 [Steady state] test3']]

# Nettoyer les données en supprimant les lignes où il manque des valeurs
df_samples_1 = df_samples_1.dropna()
df_samples_2 = df_samples_2.dropna()
df_samples_3 = df_samples_3.dropna()

# Créer une figure
plt.figure(figsize=(10, 6))

# Tracer les courbes des trois échantillons
plt.plot(df_samples_1.iloc[:, 0], df_samples_1.iloc[:, 1], label="Sample 1 (Test 1)", color='b', linewidth=2)
plt.plot(df_samples_2.iloc[:, 0], df_samples_2.iloc[:, 1], label="Sample 1 (Test 2)", color='r', linewidth=2)
plt.plot(df_samples_3.iloc[:, 0], df_samples_3.iloc[:, 1], label="Sample 1 (Test 3)", color='g', linewidth=2)

# Définir l'échelle logarithmique sur l'axe des x (temps)
plt.xscale('log')

# Ajouter les titres, les labels et la légende
plt.xlabel("Time (µs)")
plt.ylabel("Correlation Coefficient (g₂-1)")
plt.title("Autocorrelation vs Time (Logarithmic scale)")
plt.legend()

# Afficher le graphique
plt.grid(True, which="both", ls="--")  # Afficher la grille avec les ticks logarithmiques
plt.show()
```
